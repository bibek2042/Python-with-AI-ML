```python
lms = pd.read_csv("datasets/dp_lms_student_analytics_dataset.csv")
lms
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Correct month order
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

# Count student enrollments per month
monthly_enrollments = ( lms.groupby("Month").size() .reindex(months))  
# Plot line chart
plt.figure(figsize=(8,5))
plt.plot(monthly_enrollments.index, monthly_enrollments.values, color="red", marker="o")
plt.title("Monthly Student Enrollments")
plt.xlabel("Month")
plt.ylabel("Number of Students")
plt.grid(True)

plt.show()
```

```python
MonthlyRevenue = (lms.groupby("Month")["CourseFee"].sum()
    .reindex(["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"])
)

plt.figure(figsize=(8,5))
plt.plot(MonthlyRevenue.index, MonthlyRevenue.values,color="pink", marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)
plt.show()
```

```python
MonthlyProgress = ( lms.groupby("Month")["CourseProgress"].mean()
    .reindex(["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"])
)

plt.figure(figsize=(8,5))
plt.plot(MonthlyProgress.index, MonthlyProgress.values, color="purple",marker="o")
plt.title("Average Course Progress by Month")
plt.xlabel("Month")
plt.ylabel("Average Progress (%)")
plt.grid(True)
plt.show()
```

```python
MonthlyAttendance = (lms.groupby("Month")["AttendancePercentage"].mean()
    .reindex(["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"])
)

plt.figure(figsize=(8,5))
plt.plot(MonthlyAttendance.index, MonthlyAttendance.values,color="orange", marker="o")
plt.title("Average Attendance Percentage by Month")
plt.xlabel("Month")
plt.ylabel("Attendance (%)")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

course_students = lms.groupby("Course").size()

plt.figure(figsize=(11,6))
plt.bar(course_students.index, course_students.values,color="green",edgecolor="blue")

plt.title("Number of Students in Each Course",color="red")
plt.xlabel("Course", )
plt.ylabel("Number of Students")
plt.grid(axis="y",linestyle="--",alpha=1)


plt.show()
```

```python
course_revenue = lms.groupby("Course")["CourseFee"].sum()

plt.figure(figsize=(11,5))
plt.bar(course_revenue.index, course_revenue.values,color="red")

plt.title("Revenue by Each Course", fontsize=20,color="green",)
plt.xlabel("Course",fontsize=20)
plt.ylabel("Revenue",fontsize=20)


plt.show()
```

```python
avg_score = lms.groupby("Course")["FinalScore"].mean()

plt.figure(figsize=(11,5))
plt.bar(avg_score.index, avg_score.values,color="pink",edgecolor="black")

plt.title("Average Final Score by Course",fontsize=20,color="blue")
plt.xlabel("Course")
plt.ylabel("Average Final Score")

plt.show()
```

```python
avg_attendance = lms.groupby("Course")["AttendancePercentage"].mean()

plt.figure(figsize=(11,5))
plt.bar(avg_attendance.index, avg_attendance.values,color="orange")
plt.title("Average Attendance Percentage by Course",color="crimson",fontsize=20)
plt.xlabel("Course")
plt.ylabel("Attendance (%)")
plt.grid(axis="y",linestyle="--",alpha=0.1)

plt.show()
```

```python
city_students = lms.groupby("City").size()

plt.figure(figsize=(11,5))
plt.bar(city_students.index, city_students.values,color="grey")

plt.title("Number of Students from Each City",fontsize=20,color="red")
plt.xlabel("City")
plt.ylabel("Number of Students")


plt.show()
```

```python
instructor_students = lms.groupby("Instructor").size()

plt.figure(figsize=(10,5))
plt.bar(instructor_students.index, instructor_students.values,edgecolor="red")

plt.title("Number of Students under Each Instructor",fontsize=20)
plt.xlabel("Instructor")
plt.ylabel("Number of Students")
plt.grid(axis="y",linestyle="--",alpha=1)

plt.show()
```

```python
import matplotlib.pyplot as plt

top_courses = lms["Course"].value_counts().head(10)

plt.figure(figsize=(8,6))
plt.barh(top_courses.index, top_courses.values,color="green")

plt.title("Top 10 Courses by Enrollment",fontsize=20,color="red")
plt.xlabel("Number of Students",fontsize=15)
plt.ylabel("Course",fontsize=15)

plt.show()
```

```python
top_instructors = lms["Instructor"].value_counts().head(10)

plt.figure(figsize=(8,6))
plt.barh(top_instructors.index, top_instructors.values,color="pink",edgecolor="black")

plt.title("Top 10 Instructors by Student Enrollment",fontsize=20)
plt.xlabel("Number of Students",fontsize=15)
plt.ylabel("Instructor",fontsize=15)

plt.show()
```

```python
top_students = lms.sort_values(by="FinalScore", ascending=False).head(10)

plt.figure(figsize=(8,6))
plt.barh(top_students["StudentName"], top_students["FinalScore"],edgecolor="grey")

plt.title("Top 10 Students with Highest Final Scores",fontsize=25)
plt.xlabel("Final Score",fontsize=15)
plt.ylabel("Student",fontsize=15)
plt.gca().invert_yaxis() 

plt.show()
```

```python
import matplotlib.pyplot as plt

columns = [
    "CourseFee",
    "attendance_percent",
    "assignment_score",
    "quiz_score",
    "final_score",
    "course_progress"
]

titles = [
    "Course Fee",
    "Attendance Percentage",
    "Assignment Score",
    "Quiz Score",
    "Final Score",
    "Course Progress"
]

plt.figure(figsize=(15, 8))

for i in range(len(columns)):
    plt.subplot(2, 3, i + 1)
    plt.hist(lms[columns[i]], bins=10)
    plt.title(titles[i])
    plt.xlabel(titles[i])
    plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
